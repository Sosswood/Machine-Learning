### Checking Missing Data (2001 - 2021)
Let's isolate our check to only the columns from 2001 to 2021 in both datasets.

```python
# Define the target years
target_years = [str(year) for year in range(2001, 2022)]

# Filter columns that are present in our datasets
gdp_years = [col for col in target_years if col in gdp_raw.columns]
pop_years = [col for col in target_years if col in pop_raw.columns]

# Calculate missingness for GDP (2001-2021)
gdp_target_missing = gdp_raw[gdp_years].isnull().sum()
print("=== Missing GDP Data (2001-2021) ===")
if gdp_target_missing.sum() == 0:
    print("No missing data in the GDP dataset for this period!")
else:
    print("Total missing values in this range:", gdp_target_missing.sum())
    print(gdp_target_missing[gdp_target_missing > 0])

# Calculate missingness for Population (2001-2021)
pop_target_missing = pop_raw[pop_years].isnull().sum()
print("\n=== Missing Population Data (2001-2021) ===")
if pop_target_missing.sum() == 0:
    print("No missing data in the Population dataset for this period!")
else:
    print("Total missing values in this range:", pop_target_missing.sum())
    print(pop_target_missing[pop_target_missing > 0])
```

### Task A: Correlation.  Mean Population vs. Mean Per Capita GDP (2001-2021)
We will calculate the per capita GDP for each year, find the averages over the 2001-2021 period for each country, and calculate the Pearson correlation coefficient.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Filter both cleaned datasets for the overlapping years 2001-2020 (as GDP goes up to 2020)
years = [str(y) for y in range(2001, 2021)]

# Set country names as index to align them easily
gdp_aligned = gdp_clean.set_index('Country Name')[years]
pop_aligned = pop_clean.set_index('Country Name')[years]

# Find the intersection of countries available in both cleaned datasets
common_countries = gdp_aligned.index.intersection(pop_aligned.index)
gdp_aligned = gdp_aligned.loc[common_countries]
pop_aligned = pop_aligned.loc[common_countries]

# Convert both datasets to numeric types to avoid division/TypeError string issues
gdp_aligned = gdp_aligned.apply(pd.to_numeric, errors='coerce')
pop_aligned = pop_aligned.apply(pd.to_numeric, errors='coerce')

# Calculate Per Capita GDP for each year
per_capita_gdp_yearly = gdp_aligned / pop_aligned

# Calculate mean population and mean per capita GDP over the years for each country
mean_population = pop_aligned.mean(axis=1)
mean_per_capita_gdp = per_capita_gdp_yearly.mean(axis=1)

# Combine into a single DataFrame for analysis
analysis_df = pd.DataFrame({
    'Mean Population': mean_population,
    'Mean Per Capita GDP': mean_per_capita_gdp
}).dropna() # Drop any rows where calculation couldn't be performed

# Calculate correlation
correlation = analysis_df['Mean Population'].corr(analysis_df['Mean Per Capita GDP'])
print(f"Pearson Correlation Coefficient: {correlation:.4f}")

# Visualize the relationship
plt.figure(figsize=(10, 6))
sns.scatterplot(data=analysis_df, x='Mean Population', y='Mean Per Capita GDP', alpha=0.7)
plt.xscale('log')  # Use log scale since population scales span several orders of magnitude
plt.yscale('log')  # Use log scale for per capita GDP for better visualization
plt.title('Mean Population vs. Mean Per Capita GDP (2001-2020) - Log Scale')
plt.xlabel('Mean Population (Log Scale)')
plt.ylabel('Mean Per Capita GDP in USD (Log Scale)')
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.show()
```

There are no real groups or clusters here.

A Pearson correlation coefficent of -0.1387 shows that there's a very weak relationship between the mean population of each country and the mean per capital GDP.

###Task B: Regression.  
Perform linear regression, where the independent variable is the mean population of each country (from 2001 to 2021) and dependent variable is mean per capita GDP (from 2001 to 2021).

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare data for Scikit-Learn and sort by population to prevent zig-zag lines
sorted_df = analysis_df.sort_values(by='Mean Population')

X = sorted_df[['Mean Population']].values
y = sorted_df['Mean Per Capita GDP'].values

# Initialize and fit the Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Get model parameters
slope = model.coef_[0]
intercept = model.intercept_
y_pred = model.predict(X)
r2 = r2_score(y, y_pred)

print("=== Linear Regression Results ===")
print(f"Slope (Coefficient): {slope:.8f}")
print(f"Intercept: {intercept:.4f}")
print(f"R-squared Score: {r2:.4f}")

# Create two side-by-side plots: one on Linear Scale and one on Log Scale
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Left Plot: Linear Scale
sns.scatterplot(data=sorted_df, x='Mean Population', y='Mean Per Capita GDP', alpha=0.7, label='Actual Data', ax=axes[0])
axes[0].plot(sorted_df['Mean Population'], y_pred, color='red', linewidth=2, label='Regression Line')
axes[0].set_title('Linear Regression (Original Linear Scale)')
axes[0].set_xlabel('Mean Population')
axes[0].set_ylabel('Mean Per Capita GDP (USD)')
axes[0].legend()
axes[0].grid(True, ls="--", alpha=0.5)

# Right Plot: Log-Log Scale (where the regression line correctly curves as a smooth fit)
sns.scatterplot(data=sorted_df, x='Mean Population', y='Mean Per Capita GDP', alpha=0.7, label='Actual Data', ax=axes[1])
axes[1].plot(sorted_df['Mean Population'], y_pred, color='red', linewidth=2, label='Regression Line (Linear Fit)')
axes[1].set_xscale('log')
axes[1].set_yscale('log')
axes[1].set_title('Linear Regression (Log-Log Scale view)')
axes[1].set_xlabel('Mean Population (Log Scale)')
axes[1].set_ylabel('Mean Per Capita GDP in USD (Log Scale)')
axes[1].legend()
axes[1].grid(True, which="both", ls="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
