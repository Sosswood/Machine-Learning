```python

# calculate the Pearson's correlation between two variables
from numpy import mean
from numpy import std
from numpy import cov
from numpy.random import randn
from numpy.random import seed
from matplotlib import pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr
# seed random number generator
seed(1)

# prepare data
data1 = 20 * randn(1000) + 100
data2 = data1 + (10 * randn(1000) + 50)

# calculate covariance matrix
covariance = cov(data1, data2)

# calculate Pearson's correlation
corr, _ = pearsonr(data1, data2)

# plot
plt.scatter(data1, data2)
plt.show()

# summarize
print('data1: mean=%.3f stdv=%.3f' % (mean(data1), std(data1)))
print('data2: mean=%.3f stdv=%.3f' % (mean(data2), std(data2)))
print('Covariance: %.3f' % covariance[0][1])
print('Pearsons correlation: %.3f' % corr)
```

```python
from numpy import mean
from numpy import std
from numpy import cov
from numpy.random import randn
from numpy.random import seed
from matplotlib import pyplot as plt
from scipy.stats import pearsonr

# --- Modifiable Variables ---
new_seed = 2 # Change the seed to get different random numbers
num_samples = 500 # Change the number of data points
data1_std_multiplier = 30 # Change std multiplier for data1
data1_mean_offset = 120 # Change mean offset for data1
data2_noise_std_multiplier = 15 # Change std multiplier for data2 noise (larger values reduce correlation)
data2_noise_mean_offset = 60 # Change mean offset for data2 noise
# --------------------------

seed(new_seed)

# Prepare data with modified variables
data1 = data1_std_multiplier * randn(num_samples) + data1_mean_offset
data2 = data1 + (data2_noise_std_multiplier * randn(num_samples) + data2_noise_mean_offset)

# Calculate covariance matrix
covariance = cov(data1, data2)

# Calculate Pearson's correlation
corr, _ = pearsonr(data1, data2)

# Plot
plt.figure(figsize=(8, 6))
plt.scatter(data1, data2, alpha=0.7)
plt.title(f'Scatter Plot (Seed: {new_seed}, Samples: {num_samples}, Pearson Correlation: {corr:.3f})')
plt.xlabel(f'data1 (Mean: {mean(data1):.3f}, Std Dev: {std(data1):.3f})')
plt.ylabel(f'data2 (Mean: {mean(data2):.3f}, Std Dev: {std(data2):.3f})')
plt.grid(True)
plt.show()

# Summarize
print(f'data1: mean={mean(data1):.3f} stdv={std(data1):.3f}')
print(f'data2: mean={mean(data2):.3f} stdv={std(data2):.3f}')
print(f'Covariance: {covariance[0][1]:.3f}')
print(f'Pearsons correlation: {corr:.3f}')
```
